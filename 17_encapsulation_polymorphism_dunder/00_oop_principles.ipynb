## Method, Property, Classmethod, Staticmethod in Class

```python
import math


class Circle:
    DEFAULT_RADIUS = 1
    
    def __init__(self, radius: int | float):
        self.radius = radius
    
    # Usual method
    def calculate_area(self) -> float:
        return math.pi * self.radius ** 2
    
    # Accessed as property - with . notation
    # But can include any calculations
    @property
    def area(self) -> float:
        return math.pi * self.radius ** 2
    
    # Has no access to instance or class attributes
    # Can be called both with classname and instance
    @staticmethod
    def calculate_area_with_radius(radius: int | float):
        return math.pi * radius ** 2
    
    # Has access to class attributes only
    # Can be called both with classname and instance
    @classmethod
    def calculate_default_area(cls):
        return math.pi * cls.DEFAULT_RADIUS ** 2


circle = Circle(5)

print('Usual method')
print(circle.calculate_area())

print('Property access')
print(circle.area)

print('Static method')
print(Circle.calculate_area_with_radius(5))
print(circle.calculate_area_with_radius(5))

print('Class method')
print(Circle.calculate_default_area())
print(circle.calculate_default_area())
        
```

## Polymorphism

Polymorphism is an ability to use common interface for multiple data types.    
Firstly show on 01_basics_polymorphism.py

```python
class Animal:
    def __init__(self, name):
        self.name = name
    
    def say_hi(self) -> str:
        """Docstring Say hi"""
        raise NotImplementedError('Must be implemented in a child class')

class Cat(Animal):
    def say_hi(self) -> str:
        return 'meow'
    
class Dog(Animal):
    def say_hi(self) -> str:
        return 'bow'


def animals_say_hello(animal: Animal):
    print(animal.say_hi())

cat = Cat(122424)
dog = Dog('dog')
animal = Animal([1,2,3,4])
for animal in [cat, dog, animal]:
    animals_say_hello(animal)
    print(animal.name)


```

```python
animals_say_hello(cat)
animals_say_hello(dog)
```

## Encapsulation

### Not fully correct:
“Private” instance variables that cannot be accessed except from inside an object don’t exist in Python. However, there is a convention that is followed by most Python code: a name prefixed with an underscore (e.g. _spam) should be treated as a non-public part of the API (whether it is a function, a method, or a data member). It should be considered an implementation detail and subject to change without notice.

### Detailed difference:
Private fields (double underscore) - can only be accessed inside of class (implementation, etc) and are intended to be NOT VISIBLE for end user (programmer, who is going to use this class).

Protected fields (single underscore) - can be accessed inside of class, in child classes, and better not be accessed from instance by end user (although Python allows such behaviour).

Public fields (without underscore) - can be accessed and modified everywhere

```python
class Test:
    _x = 1
    __x = 2
    
    def __test(self):
        print(self.__x)
    
    def test(self):
        self.__test()

test = Test()
test.test()
test._Test__test()


```

```python
class Protected:
    
    def _protected_method(self):
        return 43
    
    def __private_method(self):
        return 1

class Child(Protected):
    def public_method(self):
        return Protected._Protected__private_method(self)

p = Protected()
print(p._protected_method())

c = Child()
print(c.public_method())
    
```

## Dunder methods

A class can implement certain operations that are invoked by special syntax (such as arithmetic operations or subscripting and slicing) by defining methods with special names. This is Python’s approach to operator overloading, allowing classes to define their own behavior with respect to language operators. For instance, if a class defines a method named __getitem__(), and x is an instance of this class, then x[i] is roughly equivalent to type(x).__getitem__(x, i). Except where mentioned, attempts to execute an operation raise an exception when no appropriate method is defined (typically AttributeError or TypeError).

```python
x = 1
y = 2
z = x + y
print(z)
```

```python
from __future__ import annotations


class MyNum:
    def __init__(self, value):
        self.value = value
        
    def __add__(self, other: MyNum) -> MyNum:
        if not isinstance(other, type(self)):
            raise ValueError('wrong type for operand')
        return MyNum(self.value + other.value)
    
    def __str__(self):
        return f"<MyNum: {self.value}>"
    
    def __repr__(self):
        return self.__str__()
```

```python
x = MyNum(1)
y = MyNum(2)
str(x)
repr(y)
```
